# Feature Engineering

```python
def create_engagement_score(df):
    df["Engagement_Score"] = (
        df["Likes"] + df["Comments"] + df["Shares"]
    )
    return df
```

```python
df = create_engagement_score(df)

df.head(3)
```

```python
def create_purchase_rate(df):
    df["Purchas_Rate"] = df["Purchases"] / df["Products_Viewed"].replace(0, 1)
    return df
```

```python
df = create_purchase_rate(df)
```

```python
df.head(3)
```

```python
def create_notification_response_rate(df):
    df["Notification_Response_Rate"] = ( 
        df["Notifications_Opened"] / df["Notifications_Received"].replace(0, 1)
    )
    return df
```

```python
df = create_notification_response_rate(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_distribution(df, column):
    sns.histplot(df[column], kde=True)
    plt.title(f"{column} Distribution")
    plt.show()

def plot_boxplot(df, column):
    sns.boxplot(x=df[column])
    plt.title(f"{column} Boxplot")
    plt.show()

plot_distribution(df, "Age")
plot_distribution(df, "Daily_Usage_Minutes")
plot_distribution(df, "Sessions_Per_Day")
plot_distribution(df, "Average_Session_Duration")

plot_boxplot(df, "Age")
plot_boxplot(df, "Daily_Usage_Minutes")
plot_boxplot(df, "Average_Session_Duration")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

corr = df.corr(numeric_only=True)

# upar wala duplicate triangle chupa do
mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(20, 16))
sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f",                  # 2 decimal tak number
    annot_kws={"size": 8},      # numbers ka chota font
    cmap="coolwarm",
    vmin=-1, vmax=1, center=0,  # -1 se +1 ka proper scale
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink": 0.7},
)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
