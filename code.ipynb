```python
categorical_features = [
    "Gender",
    "Country",
    "Device_Type",
    "Subscription",
    "Video_Used",
    "Chat_Used",
    "Shopping_Used",
    "Music_Used",
    "Search_Used",
    "Payment_Used"
]

for i in categorical_features:
    display("\n" + "="*50)
    display(f"Column : {i}")
    display(f"Unique Values : {df[i].unique()}")
    display(f"Number of Unique Values : {df[i].nunique()}")
    display(f"Value Counts {df[i].value_counts()}")
```
